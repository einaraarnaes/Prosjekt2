## Oppgave 1: Diffusjonslikningen

#### Oppgave 1a)

Vi ønsker å vise at likningen
$$\phi(x,t) = \frac{1}{\sqrt{4\pi Dt}} \int_{-\infty}^{\infty} \mathrm{d}y \, h(y) e^{-\frac{(x-y)^2}{4Dt}}$$

er en løsning at diffusjonslikningen i 1D, der D er en konstant:

$$\frac{\partial \phi(x,t)}{\partial t} = D \frac{\partial^2 \phi(x, t)}{\partial x^2}$$
Flytter først $t^{-1/2}$ inn i integralet:
$$\phi(x,t) = \frac{1}{\sqrt{4\pi D}} \int_{-\infty}^{\infty} \mathrm{d}y \, h(y) t^{-1/2}e^{-\frac{(x-y)^2}{4Dt}}$$

Antar at vi kan bytte om på integrasjons- og derivasjonsrekkefølgen. Siden det bare er den siste delen av integralet som avhenger av x og t trenger vi bare å vise at uttrykket $f(x,t) = t^{-1/2}e^{-\frac{(x-y)^2}{4Dt}}$ oppfyller likningen.
Regner først ut venstresiden:

$$\frac{\partial f}{\partial t} = -\frac{1}{2} t^{-3/2} e^{-\frac{(x-y)^2}{4Dt}} + t^{-1/2}\frac{(x-y)^2}{4Dt^2}e^{-\frac{(x-y)^2}{4Dt}}
= \left[ \frac{(x-y)^2}{4Dt} - \frac{1}{2} \right] t^{-3/2} e^{-\frac{(x-y)^2}{4Dt}}
$$

Deriverer to ganger med hensyn på x på høyresiden:
$$\frac{\partial f}{\partial x} = t^{-1/2} \frac{-2(x-y)}{4Dt} e^{-\frac{(x-y)^2}{4Dt}} = -\frac{1}{2D} t^{-3/2} \left[ (x-y) e^{-\frac{(x-y)^2}{4Dt}} \right]$$

$$\frac{\partial^2 f}{\partial x^2} = -\frac{1}{2D} t^{-3/2} \left[ e^{-\frac{(x-y)^2}{4Dt}} - \frac{2(x-y)^2}{4Dt} e^{-\frac{(x-y)^2}{4Dt}}\right] = \frac{1}{D} \left[ \frac{(x-y)^2}{4Dt} - \frac{1}{2} \right] t^{-3/2} e^{-\frac{(x-y)^2}{4Dt}}$$

Av dette ser vi lett at $\frac{\partial f}{\partial t} = D\frac{\partial^2 f}{\partial x^2}$, som var det vi ville vise.

i etterkant innser vi at dette er en konvolusjon, og at vi like gjerne kunne ha trukket frem noen gamle kunster fra matte 4K og i steden vist at fouriertransformene løser liknigen.

Fysisk tolkning:

Idet vi kom på at dette var en konvolusjon, kom vi også på at vi jo har sett dette tidligere da vi løste varmelikningen i 4K. Fra det husker vi at funksjonen $h(y)$ var initialbetingelsen vår. Det må bety at tolkningen er at $h(x)$ er fordelingen av partikler langs x-aksen i $t = 0$. Vi titter litt på grenseverdien for å forsikre oss om at teorien vår stemmer:

Hvis vi lar $t \rightarrow 0$ i $\phi(x,t)$ ser vi at eksponensialleddet i integralet går mot 0 i alle andre punkter enn $y = x$. Vi må videre anta at grenseverdien er et eller annet som gjør at integralet ikke blir null selv om integranden nærmer seg null i alle punkter med unntak av $y = x$. Uansett vil det si at vi bare burde stå igjen med $h(x)$ som det x-avhengige leddet, siden x i eksponensialleddet forsvinner. Selv om denne betraktningen blir alt for banal, gir den i alle fall en viss intuisjon for at $h(x)$ må beskrive fordelingen i $t = 0$.

#### Oppgave 1b)

Vi har at $P(x) = \frac{e^{-\beta V(x)}}{Z}$, der $\beta = (k_BT)^{-1}$. Vi lar sannsynligheten for at an partikkel beveger seg fra en posisjon x til $x_0 \in \{x-1, x, x+1\}$ er proposjonal med $P(x)$, og vi antar at denne proposjonalitetskonstanten er lik for alle mulige steg. VIl bestemme sannsynlighetene $p^+, p^0, p^-$ for at partikkelen flytter seg til høyre, blir eller til venstre. 

Vi har først kravet om at summen av de tre sannsynligghetene skal være 1 (vi kan se bort fra proposjonalitetsfaktoren).

$$\frac{e^{-\beta V(x_0 + 1)}}{Z} + \frac{e^{-\beta V(x_0)}}{Z} + \frac{e^{-\beta V(x_0 - 1)}}{Z} = 1$$
$$Z = e^{-\beta V(x_0 + 1)} + e^{-\beta V(x_0)} + e^{-\beta V(x_0 - 1)}$$

Vi får dermed:

$$ p^+ = P(x_0 + 1) = \frac{e^{-\beta V(x_0 + 1)}} {e^{-\beta V(x_0 + 1)} + e^{-\beta V(x_0)} + e^{-\beta V(x_0 - 1)}} = \frac{1}{1 + e^{-\beta[V(x_0-1)-V(x_0+1)]} + e^{-\beta[V(x_0)-V(x_0+1)]}}$$

Uttrykk for $p^0$ og $p^-$ får vi helt tilsvarende ved å regne ut $P(x_0)$ og $P(x_0-1)$, så vi skriver det ikke opp her.


#### Oppgave 1c) 

Vi ser på de to grensetilfellene for $p^+$, $p^0$ og $p^-$.

Når $k_{\mathrm{B}} T \gg |V(x+1)-V(x)|$ for alle x vil $\beta |V(x+1)-V(x)| \ll 1$. Det vil si at alle eksponensialleddene går mot $e^0 = 1$, og dermed får vi at $p^+ \approx p^0 \approx p^- \approx \frac{1}{3}$

Når $k_{\mathrm{B}} T \ll |V(x+1)-V(x)|$ for alle x vil $\beta |V(x+1)-V(x)| \gg 1$ det vil si at alle eksponensialleddene i $p^-$ går mot 0, slik at $p^- \approx 1$, mens $p_0$ og $p^+$ begge har eksponensialledd som vokser, slik at $p_0 \rightarrow 0$ og $p^+ \rightarrow 0$.

Dette synes å være fornuftig, små potensialforskjeller relativt til partiklenes energi ikke burde ha noen påvirkning på deres bevegelse, mens store potensialforskjeller burde gjøre det mer sannsynlig å bevege seg til lavere poteniale. Vi kan i disse to tilfellene forenkle virrevandringsproblemet, ved i det første å neglisjere potensialforskjellene slik at virrevandringen blir helt tilfeldig, og i det andre tilnærme det neste steget som helt sikkert. 



## Oppgave 2: Virrevandring

```python
import numpy as np
from matplotlib import pyplot as plt
import numpy.typing as npt
from collections.abc import Callable

"""
Parametere som brukes i koden med forklaring:
"""

alpha = 0.8 # Parameter som bekriver potenialet
bk = 1000 # Samlet konstant som beskriver temperaturen til partiklene. Høy bk git lav temperatur, og motsatt.
period = 100 #Perioden til potensialet, gitt som N_x i oppgaveheftet
T_p = 300 # Antall tiddsteg før potenialet byttes
size = 20 # Størrelsen på en partikkel
axis_size = 10*period # Lengden på x-aksen der partiklene oppholder seg
N_p = 100 #Antall partikler
cycles = 3 # Antall sykluser der partiklene befinner seg i de to potenialene
n_sim = 10 #Antall simuleringer
```

Funksjoner vi trenger:

```python

"""
random_walk_repulsion

Funksjonen er egnet til alle oppgavene, med unntak av 4 a) der vi skal animere og trenger posisjonene i hvert tidssteg.

Parametere:
    n_steps: Antall tidssteg som skal utføres
    start_pos: Liste med statposisjonen til alle partiklene.
    p_minus: Liste med sannsynligheter for å bevege seg mot venstre for en periode av potensialet
    p_pluss: Liste med sannsynligheter for å bevege seg mot høyre for en periode av potensialet
    repulsion: Om partiklene skal utsettes for hard repulsion eller ikke.
    axis_wrap: Om x_aksen skal være syklisk. Angis som [minste posisjon, største_posisjon], hvis ikke er x_aksen uendelig lang.
    size: Størrelsen til partiklene (kun nødvendig om repulsion = True)

Returnerer:
    pos: Numpy-array med størrelse (len(x_start)). SLuttposisjoner.
    delta_pos: Numpy array med størrelse (len(x_start)). Antall steg hver partikkel totalt har forflyttet seg.
"""
def random_walk_universal(n_steps: int, start_pos: npt.ArrayLike, p_minus: npt.ArrayLike, p_pluss: npt.ArrayLike, repulsion = False, axis_wrap: tuple = None, size: int = 1):

    n_part = len(start_pos)
    pos = np.copy(start_pos)
    delta_pos = np.zeros(n_part)
    period = len(p_pluss)

    if (axis_wrap is not None): 
        axis_start = axis_wrap[0]
        axis_end = axis_wrap[1]
        axis_length = axis_end-axis_start + 1
    else:
        axis_start = 0

    #Transformerer koordinataksen slik at den starter i 0 så moduloregning kan benyttes.
    pos = pos - axis_start

    #Om det ikke er repulsjon tar alle partiklene et steg samtidig.
    if (not repulsion):
        
        for i in range(n_steps):
            p_pos_minus = p_minus[pos % period]
            p_pos_pluss = p_pluss[pos % period]
            step = np.random.uniform(size = n_part)
            step[step <= p_pos_minus] = -1
            step[(step > p_pos_minus) & (step <= 1-p_pos_pluss)] = 0
            step[step > 1-p_pos_pluss] = 1

            pos = pos + step.astype(int)
        
        delta_pos = pos + axis_start - start_pos

        #Gjør moduloregningen, og transfprmerer tilbake koordinataksen.
        if (axis_wrap is not None):
            pos = pos%axis_length + axis_start
    

    if (repulsion):

        for i in range(n_steps):

            #Stokker om indeksene for å gjøre stegene i tilfeldig rekkefølge
            random_indexes = np.arange(n_part)
            np.random.shuffle(random_indexes)

            moves = np.zeros(n_part)

            for i in random_indexes:
                step = np.random.uniform()
                if step <= p_minus[pos[i]%period]:
                    step = -1
                elif step > (1 - p_pluss[pos[i]%period]):
                    step = 1
                else:
                    step = 0
                
                if (axis_wrap is not None):
                    #Gjør bare et steg dersom det ikke er partikkel i veien
                    if not (pos == (pos[i] + step*(size-1) + step)%axis_length).any():
                        pos[i] = (pos[i] + step)%axis_length
                        moves[i] = step
                else:
                    #Gjør bare et steg dersom det ikke er partikkel i veien.
                    if not (pos == (pos[i] + step*(size-1) + step)).any():
                        pos[i] += step
                        moves[i] = step

            delta_pos += moves.astype(int)
        
        #Transformerer tilbake koordinataksen
        if (axis_wrap is not None):
            pos = pos + axis_start
    
    return pos, delta_pos

"""
p_vals

Parametere:
    x: Liste med x-verdier.
    V: Funksjon for et potensiale som tar x som parameter.

Returnerer:
    p_pluss: Sannsynligheter for å bevege seg mot høyre.
    p_minus: Sannsynlgiheter for å bevege seg mot høyre.
"""
def p_vals(x: npt.ArrayLike, V: Callable, alpha: float, bk: float):
    p_pluss = 1 / (1 + np.exp(np.clip(-bk*(V(x-1, alpha)-V(x+1, alpha)), -700, 700)) + np.exp(np.clip(-bk*(V(x, alpha)-V(x+1, alpha)), -700, 700)))
    p_minus = 1 / (1 + np.exp(np.clip(-bk*(V(x+1, alpha)-V(x-1, alpha)), -700, 700)) + np.exp(np.clip(-bk*(V(x, alpha)-V(x-1, alpha)), -700, 700)))
    return p_minus, p_pluss
```

#### Oppgave 2a)

```python
T_p = 200
n_sim = 100
bk = 1
x_start = np.arange(-50, 51, 1)
N_p = len(x_start)

def V_1(x: npt.ArrayLike, alpha: float):
    return -x

def V_2(x: npt.ArrayLike, alpha: float):
    return np.ones_like(x)

p_minus_1, p_pluss_1 = p_vals(x_start, V_1, alpha, bk)
p_minus_2, p_pluss_2 = p_vals(x_start, V_2, alpha, bk)

walk_1 = np.zeros(shape = (n_sim, N_p))
walk_2 = np.zeros(shape = (n_sim, N_p))

for i in range(n_sim):
    positions, delta_pos = random_walk_universal(T_p, x_start, p_minus_1, p_pluss_1)
    walk_1[i] = positions
    positions, delta_pos = random_walk_universal(T_p, x_start, p_minus_2, p_pluss_2)
    walk_2[i] = positions

fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(12, 2.5))

axs[0].hist(walk_1.flatten(), bins = 200, density = True)
axs[0].set_title("V(x)=-kx")
axs[1].hist(walk_2.flatten(), bins = 200, density = True)
axs[1].set_title("V(x)=k")

for ax in axs:
    ax.set_xlabel("x")
    ax.set_ylabel("Partikkeltetthet")
```

Dette viser at partiklene fremdeles oppfører seg ganske likt som i et konstant potensial relativt til hverandre, men at hele fordelingen flytter seg. med samme hastighet.

I tilfellet med hard repulsjon ser vi at fordelingen med konstant potensial er ganske lik som tidligere. Litt overraskende er det kanskje at partikkelspredningen ser ut til å være nesten identisk som i tilfellet uten hard repulsjon, man skulle kanskje tro den ble større. Hvis vi tenker over hvordan vi har implementert den harde repulsjonen er det kanskje ikke like overraskende. Sannsynligheten for å ta et steg i hver av retningne endres ikke av at det er andre partikler rundt, så den harde repulsjonen virker ikke som et "trykk".
I det skrå potensialet ser vi at det i snitt er størst konsentrasjon av partikler ved $x=-50$, og at tettheten ser ut til å avta lineært. For å forklare dette bruker vi en halvgrei analogi med en rekke biler som står i kø og venter på grønt lys. Først står alle bilene tett i tett foran lyskrysset. Idet lyset blir grønt begynner den først bilen å kjøre, men bil 2 må vente en liten studn til den har fått litt avstand til bil 1 før den kan begynne å kjøre. Tilsvarende må bil 3 vente på bil 2, og så videre. Resultatet er jo lenger bak i køen vi kommer, jo lengre vil bilene bli stående tett i tett uten at de har beveget seg i det hele tatt. Dette forklarer fordelingen vi observerer Det er verdt å merke seg at partilene "lengst frem i køen" har beveget seg like langt som i det skrå potensialet uten hard repulsjon, omtrent til $x=175$, som er å forvente, sa disse kan bevege seg uhindret ned potensialet fra tiden $t=0$.

## Oppgave 3: Virrevandring i skrallepotensial uten interaksjoner

```python

"""
Nytt potensiale V_1
"""
def V_1(x: npt.ArrayLike, alpha: float):
    period = len(x)
    per_x = x%period-(1-alpha)*period
    return np.where(per_x > 0, per_x/(alpha*period), -per_x/((1-alpha)*period))


"""
Funksjon som gjør det litt kodebesparende å kjøre flere sykluser i deloppgavene i oppgave 3.
"""
def run_cycles(cycles, T_p, start_position, p_vals, axis_size, size = 1):
    positions = start_position.copy()

    p_minus_1, p_pluss_1, p_minus_2, p_pluss_2 = p_vals

    currents = np.zeros(cycles)
    delta_pos_2 = np.zeros(len(start_position))

    for i in range(cycles):
        positions, delta_pos_1 = random_walk_universal(T_p, positions, p_minus_1, p_pluss_1, axis_wrap = [0, axis_size-1], size = size)
        positions, delta_pos_2 = random_walk_universal(T_p, positions, p_minus_2, p_pluss_2, axis_wrap = [0, axis_size-1], size = size)
        
        currents[i] = np.mean(delta_pos_1 + delta_pos_2)/(2*T_p)
    
    return currents
```

#### Oppgave 3a)

```python
bk = 1000
period = 100
T_p = 500
N_p = 12*period
cycles = 10
axis_size = 2*period

#X-verdier for plotting
grid = np.arange(0, axis_size)

x_start = np.repeat(grid, N_p//axis_size)

x_period = np.arange(0, period)


alpha = 0.8

p_values = *(p_vals(x_period, V_2, alpha, bk)), *(p_vals(x_period, V_1, alpha, bk))
currents_1 = run_cycles(cycles, T_p, x_start, p_values, axis_size)
potential_curve_1 = V_1(x_period, alpha)[grid % period]

alpha = 0.1

p_values = *(p_vals(x_period, V_2, alpha, bk)), *(p_vals(x_period, V_1, alpha, bk))
currents_2 = run_cycles(cycles, T_p, x_start, p_values, axis_size)
potential_curve_2 = V_1(x_period, alpha)[grid % period]

with np.printoptions(precision=4, suppress=True, linewidth = 100):
    print("Gjennomsnittlig strøm for de 10 syklusene:")
    print(f"a=0.8: {currents_1}")
    print(f"a=0.1: {currents_2}")


fig, axs = plt.subplots(nrows = 1, ncols = 2, figsize=(10, 2.5))

axs[0].plot(grid, potential_curve_1)
axs[1].plot(grid, potential_curve_2)
axs[0].set_title(r"$ \alpha = 0.8$")
axs[1].set_title(r"$ \alpha = 0.1$")
for ax in axs:
    ax.set_xlabel("x")
    ax.set_ylabel("$V_1$")
plt.show()
```

Tolkning av resultatene:

* Vi får en negativ strøm som følge av at i størsteparten av området er potensialet synkende mot venstre. Dette gjør at størsteparten av partiklene strømmer mot venstre.

* La oss forsøke å forklare differansen mellom $J_{avg}(0)$ og $J_{avg}(n)$ for $n>0$. La oss starte med å anta at strømmen mens partiklene er i potensialet $V_2$ er neglisjerbar, slik at vi bare betrakter partiklenes bevegelse i $V_1$ for å forklare strømmen. I starten er alle partiklene jevnt fordelt på posisjoner fra 0 til 200, men etter de første $T_p$ stegene vil det ha oppstått en opphoping av partikler i potensialbrønnene som fører til at strømmen blir lavere i de neste syklusene. Årsaken til at strømmen ikke stopper opp helt er at mens partiklene er i potensialet $V_2$, vil noen av partiklene ha beveget seg over til andre siden av "fjelltoppen" til venstre, slik at de begynner å skli ned igjen idet de kommer tilbake igjen i $V_1$. Partiklene som har klatret oppover på hver side, men ikke har kommet seg over noen fjelltopp, vil falle tilbake igjen til samme sted og dermed ikke bidra til noen netto strøm.

* Ser nå at vi har helning andre vei, slik at strømmen går mot høyere. ```python

```python
import matplotlib.animation as animation
from IPython.display import HTML

size = 1
period = 20
axis_size = 4*period
start_pos  = np.arange(0, axis_size, 4*size)
T_p = 40
bk = 1000
alpha = 0.2
x_period = np.arange(0,period)

cycles = 5


#Funksjon for å animere
def animate(i):
    scatter.set_offsets(np.column_stack([all_positions[i], all_y_vals[i]]))
    return scatter,


"""
Funksjon som gjør random walk, men returnerer posisjonene fra hvert tidssteg slik at det kan animeres.
Laget kun for å brukes i 4a), derfor ikke like generell som random_walk_universal.
"""
def random_walk_with_animation(n_steps, x_start, p_minus, p_pluss, axis_size, size):

    period = len(p_minus)
    n_part = len(x_start)
    pos = np.copy(x_start)

    all_pos = np.zeros((n_steps, n_part), dtype = int)

    for j in range(n_steps):

        random_indexes = np.arange(n_part)
        np.random.shuffle(random_indexes)

        for k in random_indexes:
            step = np.random.uniform()
            if step <= p_minus[pos[k]%period]:
                step = -1
            elif step > (1 - p_pluss[pos[k]%period]):
                step = 1
            else:
                step = 0

            if not (pos == (pos[k] + size*step + step)%axis_size).any():
                pos[k] = (pos[k] + step)%axis_size
        
        all_pos[j] = pos.astype(int)
    
    return pos.astype(int), all_pos


p_minus_1, p_pluss_1 = p_vals(x_period, V_2, alpha ,bk)
p_minus_2, p_pluss_2 = p_vals(x_period, V_1, alpha, bk)

positions = start_pos.copy()

currents = np.zeros(cycles)

fig, ax = plt.subplots()

all_positions=np.zeros((T_p*2*cycles, len(start_pos)), dtype = int)
all_y_vals=np.zeros((T_p*2*cycles, len(start_pos)))

ax.set_ylim(-0.5, 1.5)
ax.set_xlabel("x")
ax.set_ylabel("V(x)")
ax.set_title(f"Animasjon av {len(start_pos)} partiklers bevegelse i skrallepotensialet")

scatter = ax.scatter(positions, np.ones(len(positions)))

V_2_arr = V_2(x_period, alpha)
V_1_arr = V_1(x_period, alpha)

#Kjører syklusene, og samler opp posisjoner og regner ut y_verdier tilsvarende verdien til potensialet, 
# slik at det er mulig å se hvor i potensialet paritklene befinner seg. 
for i in range(cycles):
    positions, all_positions[2*i*T_p : (2*i+1)*T_p] = random_walk_with_animation(T_p, positions, p_minus_1, p_pluss_1, axis_size, size)
    positions, all_positions[(2*i+1)*T_p : (2*i+2)*T_p] = random_walk_with_animation(T_p, positions, p_minus_2, p_pluss_2, axis_size, size)
    all_y_vals[2*i*T_p : (2*i+1)*T_p] = V_2_arr[all_positions[2*i*T_p : (2*i+1)*T_p] % period]
    all_y_vals[(2*i+1)*T_p : (2*i+2)*T_p] = V_1_arr[all_positions[(2*i+1)*T_p : (2*i+2)*T_p] % period]


ani = animation.FuncAnimation(fig, animate, frames = np.arange(len(all_positions)), interval=50, blit=True)

plt.close() # Hindrer at en ekstra statisk figur dukker opp
HTML(ani.to_jshtml())

```

Kommentarer:

Ser at når partiklene er i $V_2$ trekker de mot dalbunnen, men fordi de har en viss størrelse blir noen av dem værende ganske langt oppe. Dette burde gjøre at det er mulig å få strøm for mindre verdier av $T_p$ enn tidligere, fordi partikkelen lengst oppe i dalen ikke har like langt å vandre før den er på andre siden av fjelltoppen.

```python
alpha = 0.2
bk = 1000
period = 100
T_p = 300
size = 20
N_s = 10
axis_size = N_s*period

N_p_vals = np.linspace(1, axis_size//(size+1), 10).astype(int)

cycles = 100

densities = size*N_p_vals/(axis_size)

x_period = np.arange(0,period)

p_minus_1, p_pluss_1 = p_vals(x_period, V_2, alpha, bk)
p_minus_2, p_pluss_2 = p_vals(x_period, V_1, alpha, bk)

currents = np.zeros(len(N_p_vals))

for i, N_p in enumerate(N_p_vals):

    pos = np.linspace(0, axis_size-size, N_p).astype(int)
    
    delta_pos = np.zeros(len(pos))

    for j in range(cycles):
        pos, delta_pos_1 = random_walk_universal(T_p, pos, p_minus_1, p_pluss_1, repulsion = True, axis_wrap = [0, axis_size-1], size = size)
        pos, delta_pos_2 = random_walk_universal(T_p, pos, p_minus_2, p_pluss_2, repulsion = True, axis_wrap = [0, axis_size-1], size = size)

        delta_pos += (delta_pos_1 + delta_pos_2)
    
    currents[i] = np.mean(delta_pos)/(cycles*T_p*2)


fig, ax = plt.subplots(figsize=(5, 2.5))

ax.plot(densities, currents)
ax.set_xlabel("$Partikkeltetthet$")
ax.set_ylabel("$J_{avg}$")
ax.set_title("Sammenheng mellom partikkeltetthet og strøm")
plt.show()

```

Strømmen er positiv for alle tettheter mellom 0 og 1, som er forventet siden $a = 0.2$ og dermed har et potensiale der høyre dalside er brattest. For lav tetthet kan partiklene bevege seg tilnærmet uavhengige av hverandre, og strømmen vil være så godt som uavhengig av tettheten (siden vi beregner strøm per partikkel). Ettersom tettheten blir større, begynner interaksjonene mellom partiklene å påvirke strømmen, flere og flere partikler forsøker å flytte til optatte plasser, som fører til redusert strøm.

Som man ser er grafen ujevn for få antall partikler. Dette kan være fordi tilfeldigheter spiller en større rolle når det er få partikler, men vi er litt usikre på hvorfor vi likevel får så stor ujevnheter selv når vi kjører 100 sykluser.

Når $\rho = 1$ forventer vi at strømmen går mot null, som er det vi ser. Når alle partiklene ligger tett i tett er det ingen mulighet for bevegesle.
